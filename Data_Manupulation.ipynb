```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

print("Data:", data)

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Range:", np.max(data) - np.min(data))
print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))

```

```python
import numpy as np

marks = np.array([45, 78, 90, 55, 82, 60])

result = marks[marks > 60]

print(result)
```

```python
marks = np.array([45, 78, 90, 55, 82, 60])
marks[marks > 70]
marks[marks < 60]
marks[marks >= 60]
marks[marks == 60]
marks[marks != 60]
```

```python
marks = np.array([45, 78, 90, 55, 82, 60, 70])

result = marks[(marks >= 60) & (marks <= 80)]

print(result)
```

```python
marks = np.array([45, 78, 90, 55, 82, 60, 70])

result = marks[(marks < 50) | (marks > 80)]

print(result)
```

```python
numbers = np.array([1, 2, 3, 4, 5, 6, 7, 8])

result = numbers[numbers % 2 == 0]

print(result)
```

```python
numbers = np.array([1, 2, 3, 4, 5, 6, 7, 8])
result = numbers[numbers % 2 != 0]
print(result)
```
