```python
# ======================================================
# Concept Learning using Candidate Elimination Algorithm
# Healthy and Unhealthy Crop Identification
# ======================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------
# Step 1: Create Dataset
# ------------------------------------------------------

data = {
    'Leaf_Color' : ['Green','Green','Yellow','Brown','Green','Yellow','Brown','Green'],
    'Soil_Moisture' : ['High','Medium','Low','Low','High','Medium','Low','High'],
    'Disease_Spots' : ['No','No','Yes','Yes','No','Yes','Yes','No'],
    'Stem_Condition' : ['Strong','Strong','Weak','Weak','Strong','Weak','Weak','Strong'],
    'Crop_Status' : ['Healthy','Healthy','Unhealthy','Unhealthy',
                     'Healthy','Unhealthy','Unhealthy','Healthy']
}

df = pd.DataFrame(data)

print("Dataset")
print(df)

# ------------------------------------------------------
# Step 2: Separate Features and Target
# ------------------------------------------------------

X = np.array(df.iloc[:, :-1])
Y = np.array(df.iloc[:, -1])

# ------------------------------------------------------
# Step 3: Candidate Elimination Algorithm
# ------------------------------------------------------

def candidate_elimination(X, Y):

    specific = None

    for i, val in enumerate(Y):
        if val == "Healthy":
            specific = X[i].copy()
            break

    general = [["?" for i in range(len(specific))]
               for j in range(len(specific))]

    print("\nInitial Specific Hypothesis:")
    print(specific)

    print("\nInitial General Hypothesis:")
    print(general)

    for i, h in enumerate(X):

        if Y[i] == "Healthy":

            for x in range(len(specific)):
                if h[x] != specific[x]:
                    specific[x] = "?"
                    general[x][x] = "?"

        if Y[i] == "Unhealthy":

            for x in range(len(specific)):
                if h[x] != specific[x]:
                    general[x][x] = specific[x]
                else:
                    general[x][x] = "?"

    general = [g for g in general if g != ["?"]*len(specific)]

    return specific, general

# ------------------------------------------------------
# Step 4: Train Model
# ------------------------------------------------------

S_final, G_final = candidate_elimination(X, Y)

print("\nFinal Specific Hypothesis")
print(S_final)

print("\nFinal General Hypothesis")
for g in G_final:
    print(g)

# ------------------------------------------------------
# Step 5: Prediction Function
# ------------------------------------------------------

def predict(sample, hypothesis):

    for i in range(len(sample)):
        if hypothesis[i] == "?":
            continue
        elif sample[i] != hypothesis[i]:
            return "Unhealthy"

    return "Healthy"

# ------------------------------------------------------
# Step 6: Test New Crop
# ------------------------------------------------------

new_crop = ['Green','High','No','Strong']

print("\nNew Crop Details:")
print(new_crop)

result = predict(new_crop, S_final)

print("\nPrediction:", result)
```
